```python
import csv
import pymysql
pymysql.install_as_MySQLdb()

def load_csv_to_mysql(pfilecsv , pdelimiter=";" , phost="127.0.0.1", puser="root", ppassword="", pport=3308, pdatabase="mariadb_climsoft_db_v4" ):
    
    try:
        mydb = pymysql.connect(host=phost, user = puser, password=ppassword, port=pport, database=pdatabase)
        mycursor = mydb.cursor()

        with open(pfilecsv) as csv_file:       ####### with open('datafromclidata.csv') as csv_file:
            csvfile = csv.reader(csv_file, delimiter = pdelimiter)  ###### csvfile = csv.reader(csv_file, delimiter= ';')
            header = next(csvfile)
            all_value = []    
            for row in csvfile:
                value = (row[0], row[1], row[2], row[3], row[4], row[5],)
                
                #print("StationId", row[0])
                all_value.append(value)
        query = "INSERT INTO `DataHealthinfo` (`recordedFrom`, `describedBy`, `obsDatetime`, `obsLevel`, `obsValue`, `acquisitionType`) values (%s, %s	, %s	, %s	, %s	, %s)"
        #mycursor = mydb.cursor()
        #print(all_value)
        mycursor.executemany(query, all_value)
        mydb.commit()

        ############import MySQLdb
    except mydb.Error as err:
        print(f"Error: {err}")
        mydb.rollback()  # Crucial: Releases locks if something fails
    finally:
        mycursor.close()
        mydb.close()  # Safely close connection    
```
